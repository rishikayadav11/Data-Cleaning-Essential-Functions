1. What is data cleaning, and why is it important in data analysis?

  - What are the potential consequences of analyzing unclean or messy data?

  - Explain the common steps involved in cleaning and organizing data.

 Ans-
 **DATA CLEANING**: Data Cleaning (or Scrubbing) is the process of fixing or removing incorrect, corrupted, incorrectly formatted, duplicate, or incomplete data within a dataset.

# **Why is it important:**
   - **IMPROVES ACCURACY**: Ensures reliable analysis outcomes.
   - **SAVES TIME**: You do not waste time analysing mistakes or fixinf errors later in the process.
   - **EFFICIENCY**: Clean data allows algorithms to run faster & more reliably.
   - **ENHANCES DECISION**: Provides high-quality facts for choices.
   - **BOOSTS PRODUCTIVITY**: Helps teams focus on insights.

# **Consequences of analyzing unclean data**:
Analyzing "messy" data leads to "garbage in, garbage out", where your final results are wrong or misleading. It wastes time, causes bad business decisions, & makes machine learning models fail.

**AI DATA PREPARATION WORKFLOW** :

**Raw Data -> Data Cleaning -> Labeling -> Feature Engineering -> Split Dataset**

Cleaning data ensures your insight are accurate & reliable. It saves money, improves efficiency, & builds trust in your work by removing errors & duplicates before you start the analysis.

# **Common steps involved in cleaning and organizing data**

# **1. DATA CLEANING STEPS**:

- **REMOVE DUPLICATES RECORDS**: Eliminate identical rows to prevent skewed analysis.

- **HANDLE MISSING VALUES**: Fill gaps using averages or delete incomplete records.
- **FILTER OUT OUTLIERS**: Identify & investigate extreme, unrealistic data points.
- **VALIDATE DATA TYPES**: Ensure numbers are formatted as numeric, dates as dates & text as text.

# **2. DATA ORGANIZATION STEPS**:

- **STANDARIZED FORMATS**: Use uniform layouts, like **YYYY-MM-DD** for all dates.
- **STRUCTURE THE SCHEMA**: Arrange data into clear tables, rows, & columns.
- **NORMALIZE THE DATA**: Group related data into separate, interconnected tables.
- **CREATE A DATA DICTIONARY**: Document definitions, sources, & rules for every variable.
- **ESTABLISH VERSION CONTROL**: Track updates with clear file naming conventions.

4. What is the role of date functions like TODAY in managing datasets?

Ans- Date functions like =TODAY() are critical in data management for creating dynamic, self-updating datasets, enabling real-time analysis, calculating time durations (e.g., age or seniority), and automating workflows. These functions ensure consistency and efficiency by eliminating manual entry errors, allowing for automatic filtering, and scheduling tasks based on the current date.

**Key Roles of TODAY() in Data Management**:

- **Dynamic Time Tracking**: Because TODAY() updates every time a spreadsheet is opened or recalculated, it provides a "living" reference point for age calculation, contract expiration, or project timelines.

- **Calculating Durations**: Used to determine intervals, such as DATEDIF(Birthdate, TODAY(), "Y") to compute current age, or Deadline - TODAY() to find days remaining.

- **Automated Filtering & Reporting**: Enables automatic identification of data relevant to the present day, such as filtering for pending tasks, overdue invoices, or upcoming appointments.
- **Conditional Formatting**: Sets dynamic alerts, such as highlighting rows where a task deadline is less than or equal to TODAY().
- **Data Validation**: Ensures data entry remains relevant to the current time period.

**Examples in Data Tools**:
- **Excel/Google Sheets**: =TODAY()—returns the current date.
- **SQL Server**: GETDATE() or CAST(GETDATE() AS DATE)—used in queries to fetch records for the current date.
